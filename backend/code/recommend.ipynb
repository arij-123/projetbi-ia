```python
def clean_duplicate_names(df):
    """
    Supprime le prénom du nom de famille quand il est dupliqué
    Exemple: first_name = "Meriem", last_name = "Meriem KACHOUKH" 
             -> devient last_name = "KACHOUKH"
    """
    
    for idx, row in df.iterrows():
        first = str(row['first_name']).strip() if pd.notna(row['first_name']) else ""
        last = str(row['last_name']).strip() if pd.notna(row['last_name']) else ""
        
        # Si le prénom est présent dans le nom de famille
        if first and last and first.lower() in last.lower():
            # Supprimer le prénom du nom de famille
            last_clean = re.sub(re.escape(first), '', last, flags=re.IGNORECASE)
            # Supprimer les espaces multiples et nettoyer
            last_clean = ' '.join(last_clean.split()).strip()
            df.at[idx, 'last_name'] = last_clean
            print(f"✅ Corrigé: {first} {last} -> {first} {last_clean}")
    
    return df

# Utilisation
```

```python
df_before = df.copy()
df = clean_duplicate_names(df)
```

```python
def clean_doctor_titles(df):
    """
    Supprime les titres (Dr, Docteur, Dermatologue, Cabinet, etc.) 
    des colonnes first_name et last_name
    """
    
    # Liste des titres à supprimer
    titles = [
        'dr', 'dr.', 'docteur', 'dermatologue', 'dermato', 
        'cabinet', 'clinic', 'clinique', 'medecin', 'médecin',
        'doct', 'pr', 'prof', 'professeur', 'specialist',
        'centre', 'office', 'clinicien'
    ]
    
    for idx, row in df.iterrows():
        
        # Nettoyer first_name
        first = str(row['first_name']) if pd.notna(row['first_name']) else ""
        if first:
            first_lower = first.lower()
            for title in titles:
                if first_lower.startswith(title):
                    first = re.sub(rf'^{title}\.?', '', first, flags=re.IGNORECASE).strip()
                    break
            df.at[idx, 'first_name'] = first.title() if first else ""
        
        # Nettoyer last_name
        last = str(row['last_name']) if pd.notna(row['last_name']) else ""
        if last:
            last_lower = last.lower()
            for title in titles:
                if last_lower.startswith(title):
                    last = re.sub(rf'^{title}\.?', '', last, flags=re.IGNORECASE).strip()
                    break
            # Supprimer aussi "dermatologue" n'importe où
            last = re.sub(r'dermatologue', '', last, flags=re.IGNORECASE).strip()
            df.at[idx, 'last_name'] = last.title() if last else ""
    
    return df

```

```python
# Exécuter la fonction
df = clean_doctor_titles(df)
```

```python
def recommend_weighted_score(df, city=None, top_n=10):
    """
    Recommande avec un score pondéré:
    - Note: 70%
    - Popularité (nombre d'avis): 30%
    """
    
    filtered_df = df.copy()
    
    #if city:
    #    filtered_df = filtered_df[filtered_df['city'].str.lower() == city.lower()]
    
    filtered_df = filtered_df[filtered_df['avg_rating'] > 0]
    
    # Normaliser les notes (0-1)
    max_rating = filtered_df['avg_rating'].max()
    filtered_df['rating_norm'] = filtered_df['avg_rating'] / max_rating
    
    # Normaliser le nombre d'avis (0-1)
    max_reviews = filtered_df['rating_count'].max()
    filtered_df['reviews_norm'] = filtered_df['rating_count'] / max_reviews if max_reviews > 0 else 0
    
    # Score pondéré: 70% note + 30% popularité
    filtered_df['weighted_score'] = (filtered_df['rating_norm'] * 0.7) + (filtered_df['reviews_norm'] * 0.3)
    
    recommendations = filtered_df.sort_values('weighted_score', ascending=False).head(top_n)
    
    return recommendations[['first_name', 'last_name', 'city', 'avg_rating', 'rating_count', 'weighted_score']]

print("\n🏆 TOP 10 (Score pondéré):")
print(recommend_weighted_score(df, top_n=10))
```
